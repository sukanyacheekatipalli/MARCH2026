```python
# Create a list of first 15 Fibonacci numbers

fibonacci = [0, 1]
for i in range(2, 15):
    fibonacci.append(fibonacci[i-1] + fibonacci[i-2])
print(fibonacci)
```

```python
# Find the second largest number in [45, 67, 23, 89, 12, 34, 78]

nums = [45, 67, 23, 89, 12, 34, 78]
unique_nums = list(set(nums))
unique_nums.sort()
second_largest = unique_nums[-2]
print("Second largest number:", second_largest)
```

```python
# Create a list of prime numbers between 1 and 50

primes = []
for num in range(2, 51):
    is_prime = True
    for i in range(2, int(num**0.5) + 1):
        if num % i == 0:
            is_prime = False
            break
    if is_prime:
        primes.append(num)
print(primes)
```

```python
# Flatten nested list [[1, 2, 3], [4, 5, 6], [7, 8, 9]]

nested_list = [[1, 2, 3], [4, 5, 6], [7, 8, 9]]
flat_list = []
for sublist in nested_list:
    for item in sublist:
        flat_list.append(item)
print(flat_list)
```

```python
# Find common elements between [1, 2, 3, 4, 5] and [4, 5, 6, 7, 8]

list1 = [1, 2, 3, 4, 5]
list2 = [4, 5, 6, 7, 8]
common = list(set(list1) & set(list2))
print(common)
```

```python
# Transpose the matrix [[1, 2, 3], [4, 5, 6], [7, 8, 9]]

matrix = [[1, 2, 3],
          [4, 5, 6],
          [7, 8, 9]]
transpose = [[row[i] for row in matrix] for i in range(len(matrix[0]))]
print(transpose)
```

```python
#  Find the sum of all elements in 3D list [[[1, 2], [3, 4]], [[5, 6], [7, 8]]]
my_list = [[[1, 2], [3, 4]], [[5, 6], [7, 8]]]
sum_of_elements = sum([sum(element) for sublist in my_list for element in sublist])
print(sum_of_elements)
```
