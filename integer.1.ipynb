```python
#  Calculate the product of first 10 natural numbers
product = 1
for i in range(1, 11):
    product = product * i
print(product)
```

```python
#  Find the remainder when 156 is divided by 7
num = 156
div = 7
remainder = num % div
print(remainder)
```

```python
#  Calculate the square of 25
num = 25
square = num * num
print(square)
```

```python
#  Find the cube root of 125
num = 125
cube_root = num ** (1/3)
print(cube_root)
```

```python
#  Calculate the sum of digits in number 12345
num = 12345
print(sum(int(d) for d in str(num)))
```

```python
# Check if 97 is a prime number
num = 97
for i in range(2, num):
    if num % i == 0:
        print("Not Prime")
        break
else:
    print("Prime")
```

```python
# Find the factorial of 8
num = 8
fact = 1
for i in range(1, num + 1):
    fact *= i
print(fact)
```

```python
#  Calculate the average of numbers: 15, 23, 31, 42, 56
numbers = [15, 23, 31, 42, 56]
average = sum(numbers) / len(numbers)
print(average)
```

```python
# Find the greatest common divisor (GCD) of 48 and 36
import math
print(math.gcd(48, 36))
```

```python
# Calculate the sum of first 20 odd numbers
n = 20
sum_odd = 0
for i in range(n):
    sum_odd += (2 * i + 1)
print(sum_odd)
```
