```python
#Convert 'hello world' to 'hElLo WoRlD' (alternating case)
text = "hello world"
result = ""
for i in range(len(text)):
    if i % 2 == 0:
        result += text[i].lower()
    else:
        result += text[i].upper()
print(result)
```

```python
# Check if string is a valid email format: 'user@example.com'
import re
email = "user@example.com"
pattern = r'^[\w\.-]+@[\w\.-]+\.\w+$'
if re.match(pattern, email):
    print("Valid email")
else:
    print("Invalid email")
```

```python
# Check if string is a valid phone number: '+1-555-123-4567'
import re
phone = "+1-555-123-4567"
pattern = r'^\+\d{1}-\d{3}-\d{3}-\d{4}$'
if re.match(pattern, phone):
    print("Valid phone number")
else:
    print("Invalid phone number")
```

```python
#  Convert "snake_case" to "camelCase"
text = "snake_case"
words = text.split(' _ ')
camel_case = words[0] + ''.join(word.capitalize() for word in words[1:])
print(camel_case)
```

```python
#  Check if string is a valid palindrome ignoring case: "A man a plan a canal Panama"
text = "A man a plan a canal Panama"
cleaned = text.replace(" ", "").lower()
if cleaned == cleaned[::-1]:
    print("Palindrome")
else:
    print("Not Palindrome")
```

```python
# Find the most common word in 'the quick brown fox jumps over the lazy dog'

text = "the quick brown fox jumps over the lazy dog"
words = text.split()
most_common = max(set(words), key=words.count)
print(most_common)
```
